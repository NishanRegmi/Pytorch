**Tensor Operation in GPU**

```python
if torch.cuda.is_available():
  print("GPU is available")
  print(f"using GPU: {torch.cuda.get_device_name(0)}")
else:
  print("GPU not available. CPU is using")
```

```python
device = torch.device('cuda')
```

```python
#creating new tensor on gpu
torch.rand(2, 3, device=device)
```

```python
a = torch.rand(2, 3)
a
```

```python
#moving existing tensor to gpu
b = a.to(device)
print(b)
```

```python
import torch
import time

matrix1 = torch.randn(10000, 10000)
matrix2 = torch.randn(10000, 10000)

#measure time on cpu
start = time.time()
result_cpu = torch.matmul(matrix1, matrix2)
cpu_time = time.time()-start

print(f"Time taken on cpu: {cpu_time:.4f} seconds")

#measure time on gpu
matrix1_gpu = matrix1.to(device)
matrix2_gpu = matrix2.to(device)

start_gpu = time.time()
result_gpu = torch.matmul(matrix1_gpu, matrix2_gpu)
torch.cuda.synchronize()
gpu_time = time.time()-start_gpu

print(f"Time taken on gpu: {gpu_time:.4f} seconds")

print("\n Speedup (cpu time / gpu time)", cpu_time/gpu_time)
```
