```python
import torch
if torch.cuda.is_available():
  print("GPU is using")
  print(f"GPU model {torch.device('cuda')}")
else:
  print("CPU is working")
```

```python
"""
autograd is a core component of PyTorch that provides automatic differentiation for tensor operations. It enables gradient computation, which is
essential for training machine learning and deep learning models using optimization algorithms like gradient descent.
"""
#computing gradinet using pytorch
x = torch.tensor(5.0, requires_grad=True)
x
```

```python
y = x ** 2
y
```

```python
y.backward() #tracking back
```

```python
x.grad #this is derivative value
```

```python
#another one
x = torch.tensor(5.0, requires_grad=True)
y = x ** 2
z = torch.sin(y)
```

```python
print(x)
print(y)
print(z)
```

```python
z.backward() #backtrack
```

```python
x.grad
```

```python
#vector
x = torch.tensor([2.0, 5.0, 9.0], requires_grad=True)
x
```

```python
y = (x**2).mean()
y
```

```python
y.backward()
```

```python
x.grad #we compute the three
```

```python
x.grad.zero_() #this ensure the gradient doesnot accumulate in the loop process
```

```python
#closing the gradient tracking
x.requires_grad_(False) #underline will make inplace change
```

```python
#using detach function
z = x.detach()
z
```

```python
#using no grad function
x = torch.tensor(5.0, requires_grad=True)
x
```

```python
with torch.no_grad():
  y=x**2
```

```python
y
```

```python
y.backward() #returns error
```
