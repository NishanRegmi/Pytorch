**NN Module**

The torch.nn module in PyTorch is a core library that provides a wide array of classes and functions designed to help developers build neural networks efficiently and effectively.

Key component of torch.nn

1. Modules(layers):
2. Activatio function:
3. Loss Functions:
4. Container Modules
5. Regularization and dropout

```python
import torch
import torch.nn as nn
```

```python
class Model(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.linear = nn.Linear(num_features, 1)
    self.sigmoid = nn.Sigmoid()

  #forward pass
  def forward(self, features):
    out = self.linear(features)
    out = self.sigmoid(out)

    return out
```

```python
#create dataset
features = torch.rand(10, 5)

#create model
model = Model(features.shape[1])

#call model for forward pass
#model.forward(features) #this also works
model(features)
```

Creating new neural network with more hidden layer.

```python
import torch
import torch.nn as nn
```

```python
class NNmodule(nn.Module):

  def __init__(self, num_features):

    super().__init__()
    self.linear1 = nn.Linear(num_features, 3)
    self.relu = nn.ReLU()

    self.linear2 = nn.Linear(3, 1)
    self.sigmoid = nn.Sigmoid()

  def forward(self, features):
    out = self.linear1(features)
    out = self.relu(out)
    out = self.linear2(out)
    out = self.sigmoid(out)
    return out
```

```python
#create dataset
features = torch.rand(10, 5)

#create model
model = NNmodule(features.shape[1])

model(features)
```

```python
#weight
print(model.linear1.weight)
print(model.linear1.bias)
print(model.linear2.weight)
print(model.linear2.bias)
```

```python
from torchinfo import summary
summary(model, input_size=(10, 5))
```

Same architecture using the container model

```python
import torch
import torch.nn as nn
```

```python
from torch.nn.modules.linear import Linear
from torch.nn.modules.activation import ReLU
class nnModule(nn.Module):
  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  def forward(self, features):
    out = self.network(features)

    return out
```

```python
#create dataset
features = torch.rand(10, 5)

#create model
model = nnModule(features.shape[1])

model(features)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
df = pd.read_csv("/content/sample_data/breast_cancer.csv")
```

```python
df.head()
```

```python
df.drop(columns=['id', 'Unnamed: 32'], inplace=True)
```

```python
df.isnull().sum()
```

```python
#split into train test
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:, 1:], df.iloc[:, 0], test_size=0.2)
```

```python
print(X_train.shape)
print(y_train.shape)
```

```python
#normalize data
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_test
```

```python
#use label encoder for output column
encoder = LabelEncoder()
y_train = encoder.fit_transform(y_train)
y_test = encoder.transform(y_test)
```

```python
y_train
```

```python
import torch
import torch.nn as nn
```

```python
#conver it into tensor
X_train_tensor = torch.from_numpy(X_train).float()
X_test_tensor = torch.from_numpy(X_test).float()
y_train_tensor = torch.from_numpy(y_train).float()
y_test_tensor = torch.from_numpy(y_test)
```

```python
X_train_tensor
```

```python
#defining model
class BreastCancer(nn.Module):

  def __init__(self, num_features):
    super().__init__()
    self.network = nn.Sequential(
        nn.Linear(num_features, 5),
        nn.ReLU(),
        nn.Linear(5, 3),
        nn.ReLU(),
        nn.Linear(3, 1),
        nn.Sigmoid()
    )

  #forward pass
  def forward(self, features):
    out = self.network(features)
    return out
```

```python
#parameters
learing_rate = 0.1
epochs = 25
```

```python
loss_function = nn.BCELoss()
```

```python
#create model
model = BreastCancer(X_train_tensor.shape[1])

#define optimizer
optimizer = torch.optim.SGD(model.parameters(), lr=learing_rate)

#define loop
for epoch in range(epochs):
  #forward pass
  y_pred = model(X_train_tensor)

  #loss function
  loss = loss_function(y_pred, y_train_tensor.view(-1, 1))

  #clear
  optimizer.zero_grad()

  #backward pass
  loss.backward()

  #optimize
  optimizer.step()

  #loss in each epoch
  print(f"Epoch: {epoch+1}, Loss: {loss.item()}")
```

```python
#evaluation
with torch.no_grad():
  y_pred = model.forward(X_test_tensor)
  y_pred = (y_pred > 0.5).float()
  accuracy = (y_pred == y_test_tensor.float()).float().mean()
  print(f"Accuracy: {accuracy.item()}")
```
